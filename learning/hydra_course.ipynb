# Module 1 — Basics
## Lesson 1.1 — Your first Hydra app

### The idea 💡
Instead of hard-coding `lr = 0.001` inside your script, you put settings in a **YAML file**, and
Hydra hands them to your function as one object called `cfg`.

A Hydra app is just two things:
1. a **config file** (`conf/config.yaml`) — the settings
2. a **function** decorated with `@hydra.main` — your code, which receives `cfg`

In Colab we write files with the `%%writefile` magic, then run them with `!python`.

### Step 1 — Make a folder to work in
`%mkdir` creates the folders. `%cd` moves into one **and stays there** for later cells.

### Step 2 — Write the config
`%%writefile <path>` saves the whole cell into a file (it must be the **first line** of the cell).

### Step 3 — Write the app
What each piece does:
- `@hydra.main(...)` — "run this function with a config loaded for me"
  - `config_path="conf"` → *which folder* holds the YAML (relative to this `.py` file)
  - `config_name="config"` → *which file* (no `.yaml` needed)
  - `version_base="1.3"` → pins Hydra's behavior so upgrades don't surprise you
- `cfg` — your settings. Read values with dots (`cfg.lr`) or brackets (`cfg["epochs"]`)
- `OmegaConf.to_yaml(cfg)` — prints the config prettily

### Step 4 — Run it 🚀
Notice: you call `main()` with **no arguments** — Hydra builds `cfg` and passes it in.

### What just happened?
1. Hydra looked in `conf/` for `config.yaml` and loaded it.
2. It turned the YAML into the `cfg` object.
3. It called your `main(cfg)`.
